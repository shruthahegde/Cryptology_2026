```python
def encrypt(text, key):
    cols = len(key)
    rows = (len(text) + cols - 1) // cols
    table = []
    for i in range(rows):
        table.append(text[i * cols:(i + 1) * cols])
    order = sorted(range(cols), key=lambda i: key[i])
    cipher = ""
    for col in order:
        for row in table:
            if col < len(row):
                cipher += row[col]
    return cipher
def decrypt(cipher, key):
    cols = len(key)
    rows = (len(cipher) + cols - 1) // cols
    rem = len(cipher) % cols
    order = sorted(range(cols), key=lambda i: key[i])
    table = [[""] * cols for i in range(rows)]
    pos = 0
    for col in order:
        length = rows if rem == 0 or col < rem else rows - 1
        for row in range(length):
            table[row][col] = cipher[pos]
            pos += 1
    plain = ""
    for row in table:
        plain += "".join(row)
    return plain
text = input("Enter plaintext: ").replace(" ", "").upper()
key = input("Enter word key: ").upper()
cipher = encrypt(text, key)
print("Encrypted text:", cipher)
plain = decrypt(cipher, key)
print("Decrypted text:", plain)
```

```python
def encrypt(text, key):
    cols = len(key)
    table = []
    for i in range(0, len(text), cols):
        table.append(text[i:i+cols])
    order = []
    for ch in sorted(key):
        order.append(key.index(ch))
    cipher = ""
    for col in order:
        for row in table:
            if col < len(row):
                cipher += row[col]
    return cipher
def decrypt(cipher, key):
    cols = len(key)
    rows = (len(cipher) + cols - 1) // cols
    rem = len(cipher) % cols
    table = [[""] * cols for i in range(rows)]
    order = []
    for ch in sorted(key):
        order.append(key.index(ch))
    pos = 0
    for col in order:
        length = rows if rem == 0 or col < rem else rows - 1
        for row in range(length):
            table[row][col] = cipher[pos]
            pos += 1
    plain = ""
    for row in table:
        for ch in row:
            plain += ch
    return plain
text = input("Enter plaintext: ").replace(" ", "").upper()
key1 = input("Enter first word key: ").upper()
key2 = input("Enter second word key: ").upper()
c1 = encrypt(text, key1)
c2 = encrypt(c1, key2)
print("\nFirst encryption:", c1)
print("Final ciphertext:", c2)
d1 = decrypt(c2, key2)
d2 = decrypt(d1, key1)
print("\nFirst decryption:", d1)
print("Final plaintext:", d2)
```
