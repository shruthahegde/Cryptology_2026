```python
p= int(input("enter a prime number:"))
q=int(input("enter prime number:"))
n=p*q
phi=(p-1)*(q-1)
e=int(input("enter private key:"))
for i in range (1,phi):
 if(e*i)%phi==1:
    d=i
    break
print("public key :",e,n)
print("private key ",d,n)
message=int(input("enter the message:"))
cipher=(message**e)%n
print("encrypted message:",cipher)
plain=(cipher**d)%n
print("decrypter message:",plain)
```
